```python
def selection_sort(items):
    for step in range(len(items)):

        location_of_smallest = step
        for location in range(step, len(items)):
            if items[location] < items[step]:
                items[step], items[location] = items[location], items[step]
    
    return items

items = [5, 40, 3, 2, 1]
items = selection_sort(items)
print(items)
```

## Exercise 2.2

```python
def merge(left, right):
    result = []
    i = j = 0
    # left = [5, 6] and right = [4, 10]
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

def merge_sort(items):
    if len(items) <= 1:
        return items

    mid = len(items) // 2
    left = items[:mid]
    right = items[mid:]

    sorted_left = merge_sort(left)
    sorted_right = merge_sort(right)

    return merge(sorted_left, sorted_right)


unsorted_arr = [4,2,5,2,7,23, 75, 21, 18]
sorted_arr = merge_sort(unsorted_arr)
print(f'Merge sort: {sorted_arr}')
```

## Exercise 2.3

```python
def insertion_sort(nums):
    for i in range(len(nums)):
        val = nums[i]
        j = i - 1
        while j >= 0 and val < nums[j]:
            nums[j + 1] = nums[j]
            j -= 1
        nums[j + 1] = val
    return nums

unsorted_arr = [1, 2, 9, 6, 1]
sorted_arr = insertion_sort(unsorted_arr)
print(f'Insertion sort: {sorted_arr}')
```

```python
def bubble_sort(nums):
    for i in range(len(nums)):
        for j in range(len(nums) - 1):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]
    return nums
    
unsorted_arr = [1, 2, 9, 6, 10]
sorted_arr = bubble_sort(unsorted_arr)
print(f'Bubble sort: {sorted_arr}')          
```

```python
unsorted_arr = ['P', 'Y', 'T', 'H', 'O', 'N']
sorted_arr = bubble_sort(unsorted_arr)
print(f'Bubble sort: {sorted_arr}')  

unsorted_arr = ['P', 'Y', 'T', 'H', 'O', 'N']
sorted_arr = insertion_sort(unsorted_arr)
print(f'Insertion sort: {sorted_arr}') 

unsorted_arr = ['P', 'Y', 'T', 'H', 'O', 'N']
sorted_arr = merge_sort(unsorted_arr)
print(f'Merge sort: {sorted_arr}') 

unsorted_arr = ['P', 'Y', 'T', 'H', 'O', 'N']
sorted_arr = selection_sort(unsorted_arr)
print(f'selection sort: {sorted_arr}') 
```

## Exercise 2.4

```python
def mean(nums):
    return sum(nums) / len(nums)

def median(nums):
    sorted_nums = insertion_sort(nums)
    return sorted_nums[len(sorted_nums) // 2]

def mode(nums):
    nums_dict = dict.fromkeys(nums, 0)

    for num in nums:
        nums_dict[num] += 1
    
    max_key = 0
    for num in nums_dict.keys():
        if max_key < nums_dict[num]:
            max_key = num
    return max_key

nums = [1, 1, 2, 2, 2, 3, 4, 6]
print(f"Mean = {mean(nums)}")
print(f"Median = {median(nums)}")
print(f"Mode = {mode(nums)}")
```

## Exercise 2.5

```python
def duplicates(nums):
    nums_dict = dict.fromkeys(nums, 0)

    for num in nums:
        nums_dict[num] += 1

    dups = []
    for num in nums_dict.keys():
        if nums_dict[num] > 1:
            dups.append(num)
            
    return dups

nums = [1, 1, 2, 2, 2, 3, 4, 6]
print(f'Duplicates: {duplicates(nums)}')
```

## Exercise 2.9

```python
def linear_search(nums, target):
    for num in nums:
        if target == num:
            return True

    return False
```

```python
import time

def get_time_taken(nums, target):
    start_time = time.time()
    if linear_search(nums, target):
        end_time = time.time()
    else:
        return None

    return end_time - start_time

big_list = list(num for num in range(100000, 300001))

print(f"Time taken to find 100,000 = {get_time_taken(big_list, 100000)}")
print(f"Time taken to find 200,000 = {get_time_taken(big_list, 200000)}")
print(f"Time taken to find 300,000 = {get_time_taken(big_list, 300000)}")
print(f"Time taken to find 3 = {get_time_taken(big_list, 3)}")
```
