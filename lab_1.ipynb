```python
flavors = ["vanilla", "chocolate", "strawberry", "pistacchio"]
sauces = ["caramel", "butterscotch", "chocolate"]

def sundaes():
    total = 0
    for flavor in flavors:
        for sauce in sauces:
            print(flavor + " ice cream sundae with " + sauce + " sauce")
            total += 1
    return total
print(sundaes())
```

```python
def triangle(limit):
    value = 1
    row = 1
    while row <= value:
        if value > limit:
            break
        column = 1
        while column <= row:
            if column != row:
                print(value, ' ', sep = '', end = '')
            else:
                print(value)
            value = value + 1
            column = column + 1
        row = row + 1
        

triangle(100)
```

## Exercise 1.4.1

```python
def cube(n):
    return n * n * n
cube(3)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

factorial(5)
```

```python
def convert_list_to_string(lst):
    str = ''
    for c in lst:
        str+= c
    return str

def count_pattern(pattern, lst):
    pattern = convert_list_to_string(pattern)
    lst = convert_list_to_string(lst)

    count = 0
    for i in range(len(lst)):
        str = lst[i:]
        if len(str) < len(pattern):
            break
        flag = True
        for j in range(len(pattern)):
            if pattern[j] != str[j]:
                flag = False

        if not flag:
            count += 1

    return count

count_pattern(['a', 'b', 'a'], ['g', 'a', 'b', 'a', 'b', 'a','b', 'a'])
```

```python
def print_table(n):
    for i in range(1, 11):
        print(f"{n} x {i} = {i * n}")

print_table(5)
```

```python
def calculator(n1, n2, op):
    if op == '+':
        return n1 + n2
    elif op == '-':
        return n1 - n2
    elif op == '/':
        if n1 == 0 or n2 == 0:
            print("Division by Zero error")
            return 0
        else:
             return n1 / n2
    elif op == '*':
        return n1 * n2
    else:
        print("invalid operator")
        return None

print(calculator(5, 2, '*'))
```

```python
def convert_to_string(lst):
    string = ''
    for c in lst:
        string += c
    return string

def bubble_sort(string):
    chars = list(string)
    for i in range(len(chars)):
        for j in range(len(chars) - 1):
            if chars[j] > chars[j + 1]:
                chars[j], chars[j + 1] = chars[j + 1], chars[j]

    return convert_to_string(chars)
    

print(bubble_sort('dcba'))
```
